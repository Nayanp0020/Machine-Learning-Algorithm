```python
import numpy as np
import random
```

```python
bandits = [0.2, 0.5, 0.75, 0.9]

print("4-Armed Bandit Created")
print(bandits)
```

```python
def pull_arm(bandit_probability):

    if random.random() < bandit_probability:
        return 1 #reward
    else:
        return 0 #no reward
```

```python
reward = pull_arm(bandits[2])

print("Reward:", reward)
```

```python
num_arms = 4

rewards = np.zeros(num_arms)

counts = np.zeros(num_arms)

print("Agent Created")
```

```python
epsilon = 0.2

episodes = 1000

for episode in range(episodes):

    # Explore
    if random.random() < epsilon:
        action = random.randint(0,3)

    # Exploit
    else:
        action = np.argmax(rewards)

    reward = pull_arm(bandits[action])

    counts[action] += 1

    rewards[action] += (reward - rewards[action]) / counts[action]
```

```python
print("Learned Rewards:")
print(rewards)

best_arm = np.argmax(rewards)

print("\nBest Arm Found:", best_arm)
```
