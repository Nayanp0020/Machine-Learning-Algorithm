DBSCAN Clustering

```python
import numpy as np
import matplotlib.pyplot as plt

from sklearn.datasets import make_blobs
from sklearn.cluster import DBSCAN
```

```python
X, y = make_blobs(
    n_samples=300,
    centers=3,
    cluster_std=0.7,
    random_state=42
)

plt.scatter(X[:,0], X[:,1])
plt.title("Original Dataset")
plt.show()
```

```python
#Apply DBSCAN
dbscan = DBSCAN(
    eps=0.8,
    min_samples=5
)

labels = dbscan.fit_predict(X)
```

```python
#Display Cluster Labels
print(labels[:20])
```

```python
plt.scatter(
    X[:,0],
    X[:,1],
    c=labels,
    cmap='rainbow'
)

plt.title("DBSCAN Clustering")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")

plt.show()
```
